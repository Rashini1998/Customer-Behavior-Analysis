```python
import pandas as pd

df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.info()
```

```python
df.describe(include='all')  #to get summary statistics of all numerical & categorical columns
```

```python
#checking missing values
df.isnull().sum()
```

```python
# Mean --> get affected by outliers. That would skewed the distribution of our data.So me chose median to replace missing values
# We will find out median review rating within each category and replace with the missing value

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# Putting column names in snake casing (eg,; customer_id)

df.columns = df.columns.str.lower() # make all letters to lower case
df.columns = df.columns.str.replace(' ', '_') # replace spaces with underscores
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

### Feature Engineering and Create a new column

```python
# create a new column age_group

labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create column purchase_frequency_days

frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,     
    'Bi-Weekly' : 1,
    'Annually' : 365,
    'Every 3 Months' : 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```
